# §1.1 微分可能多様体 - 正規分布を例に理解する

情報幾何における「多様体」の最も身近な例は、確率分布のパラメータ空間です。
正規分布 $N(\mu, \sigma^2)$ のパラメータ空間は2次元多様体を形成します。

## 既知概念との対応

| 情報幾何の概念 | 既知概念 |
|--------------|--------|
| 多様体 | パラメータ空間 $(\mu, \sigma)$ |
| 多様体上の点 | パラメータの組、つまり一つの確率分布 |
| 局所座標 | パラメータ化の方法 |
| 座標変換 | 異なるパラメータ化間の変換 |

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams['figure.figsize'] = (10, 6)
```

```python
def gaussian_pdf(x, mu, sigma):
    """正規分布の確率密度関数"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
```

---
## 1. パラメータ空間を多様体として可視化

### 核心的な理解

- **各点 $(\mu, \sigma)$ は一つの正規分布に対応**
- この平面全体が「統計的多様体」
- $\sigma > 0$ の制約があるため、上半平面のみ（境界を含まない開集合）

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左図：パラメータ空間（多様体）
ax1 = axes[0]
mu_range = np.linspace(-3, 3, 20)
sigma_range = np.linspace(0.1, 3, 20)

# グリッド点をプロット
for mu in mu_range[::2]:
    for sigma in sigma_range[::2]:
        ax1.plot(mu, sigma, 'b.', markersize=3, alpha=0.5)

# いくつかの特定の分布を強調
special_points = [
    (0, 1, 'r', 'Standard normal'),
    (1, 0.5, 'g', 'Narrow distribution'),
    (-1, 2, 'm', 'Wide distribution'),
]

for mu, sigma, color, label in special_points:
    ax1.plot(mu, sigma, 'o', color=color, markersize=10, label=label)

ax1.set_xlabel('μ (mean)', fontsize=12)
ax1.set_ylabel('σ (std dev)', fontsize=12)
ax1.set_title('Parameter space = Statistical manifold\nEach point corresponds to one probability distribution', fontsize=11)
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3, label='σ=0 boundary (not included)')
ax1.legend(loc='upper right')
ax1.set_xlim(-4, 4)
ax1.set_ylim(-0.5, 4)
ax1.grid(True, alpha=0.3)

# 右図：対応する確率密度関数
ax2 = axes[1]
x = np.linspace(-6, 6, 200)

for mu, sigma, color, label in special_points:
    y = gaussian_pdf(x, mu, sigma)
    ax2.plot(x, y, color=color, linewidth=2, label=label)

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('p(x)', fontsize=12)
ax2.set_title('Corresponding probability density functions', fontsize=11)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 📝 学習ポイント

1. **多様体とは**: パラメータの組が張る空間
2. **正規分布の場合**: $(\mu, \sigma)$ の2次元空間（$\sigma > 0$）
3. **多様体上の各点は、一つの確率分布に対応**
4. この視点が情報幾何の出発点

---
## 2. 局所座標系の概念

同じ点を異なるパラメータ化で表現できます：

| 座標系 | パラメータ | 用途 |
|-------|----------|------|
| 座標系1 | $(\mu, \sigma)$ | 自然な座標 |
| 座標系2 | $(\mu, \sigma^2)$ | 分散を使う座標 |
| 座標系3 | $(\mu, 1/\sigma^2)$ | 精度を使う座標（ベイズ推定で頻出） |

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 同じ分布族を異なる座標系で表示
mu_vals = np.linspace(-2, 2, 10)
sigma_vals = np.linspace(0.5, 2, 10)

# 座標系1: (μ, σ)
ax1 = axes[0]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax1.plot(mu, sigma, 'b.', markersize=4)
ax1.set_xlabel('μ')
ax1.set_ylabel('σ')
ax1.set_title('Coordinates 1: (μ, σ)')
ax1.grid(True, alpha=0.3)

# 座標系2: (μ, σ²)
ax2 = axes[1]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax2.plot(mu, sigma**2, 'g.', markersize=4)
ax2.set_xlabel('μ')
ax2.set_ylabel('σ² (variance)')
ax2.set_title('Coordinates 2: (μ, σ²)')
ax2.grid(True, alpha=0.3)

# 座標系3: (μ, 1/σ²) - 精度パラメータ
ax3 = axes[2]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax3.plot(mu, 1/sigma**2, 'r.', markersize=4)
ax3.set_xlabel('μ')
ax3.set_ylabel('τ = 1/σ² (precision)')
ax3.set_title('Coordinates 3: (μ, τ)\nCommon in Bayesian inference')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 📝 学習ポイント

1. **同じ多様体を異なる座標系で記述できる**
2. ベイズ推定では精度 $\tau = 1/\sigma^2$ を使うことが多い
   - 事前分布と尤度の積が計算しやすい
3. **情報幾何では座標系に依存しない「不変な」量を扱う**
   - Fisher情報行列 → リーマン計量
   - KLダイバージェンス → ダイバージェンス関数

## 3. なぜ「多様体」という言葉を使うのか

### 多様体の定義（直感的）

**多様体** = 局所的にユークリッド空間のように見える空間

### 例：球面
- 球面は3D空間に埋め込まれた2次元多様体
- 局所的には平面に見える（地球上の小さな領域を考えよ）

### 統計的多様体の「曲がり」
- パラメータ空間 $(\mu, \sigma)$ 自体は平面的に見える
- しかし **Fisher計量を入れると曲がった空間になる**
- $\sigma$ が小さい領域は「Fisher的に遠い」

```python
fig = plt.figure(figsize=(14, 5))

# 左図：球面（2次元多様体の典型例）
ax1 = fig.add_subplot(121, projection='3d')
u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))
ax1.plot_surface(x, y, z, alpha=0.6, cmap='viridis')
ax1.set_title('Sphere: 2D manifold in 3D space\nLocally looks flat', fontsize=11)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('z')

# 右図：正規分布の多様体を「曲がった空間」として描画
ax2 = fig.add_subplot(122)

# ユークリッド的なグリッド
mu_range = np.linspace(-2, 2, 11)
sigma_range = np.linspace(0.3, 2, 11)

# 通常のグリッド（薄い線）
for mu in mu_range:
    ax2.plot([mu, mu], [0.3, 2], 'b-', alpha=0.2)
for sigma in sigma_range:
    ax2.plot([-2, 2], [sigma, sigma], 'b-', alpha=0.2)

# Fisher計量による「等距離線」（概念的）
# σが小さいほど、μ方向の距離が大きくなることを示唆
for i, sigma in enumerate([0.5, 1.0, 1.5]):
    # σが小さいほど密なグリッド
    mu_dense = np.linspace(-2, 2, int(20/sigma))
    for mu in mu_dense:
        ax2.plot(mu, sigma, 'r.', markersize=3)

ax2.annotate('Small σ region:\n"Fisher-far"',
             xy=(0, 0.5), xytext=(1.5, 1.2),
             arrowprops=dict(arrowstyle='->', color='red'),
             fontsize=10, color='red')

ax2.set_xlabel('μ', fontsize=12)
ax2.set_ylabel('σ', fontsize=12)
ax2.set_title('Gaussian manifold\n"Curved" with Fisher metric', fontsize=11)
ax2.set_xlim(-2.5, 2.5)
ax2.set_ylim(0, 2.5)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 📝 学習ポイント

1. **多様体** = 局所的にユークリッド空間のように見える空間
2. 球面は3D空間内の2D多様体（曲がっている）
3. **統計的多様体も「Fisher計量」を入れると曲がった空間になる**
4. $\sigma$ が小さい領域は「Fisher的に」遠い
   - 小さい $\sigma$ の分布を少し変えると大きな情報変化
   - **カルマンフィルタで観測精度が高いほど更新が大きいのと同じ直感**

## 4. 確認問題

### Q1. 多様体の次元
以下の統計的多様体の次元を答えよ：
- (a) 正規分布 $N(\mu, \sigma^2)$（$\mu, \sigma$ 共に自由）
- (b) 標準正規分布 $N(0, 1)$
- (c) 多次元正規分布 $N(\mu, \Sigma)$ where $\mu \in \mathbb{R}^d$, $\Sigma$ は $d \times d$ 正定値行列

### Q2. 座標変換
座標系 $(\mu, \sigma)$ から $(\mu, \tau = 1/\sigma^2)$ への座標変換のヤコビ行列を計算せよ。

### Q3. 直感の確認
なぜ $\sigma$ が小さい領域は「情報的に遠い」と言えるのか、カルマンフィルタの経験から説明せよ。

```python
# Q1の解答
print("""
Q1 解答:
(a) N(μ, σ²): 2次元 (μとσの2つのパラメータ)
(b) N(0, 1): 0次元 (パラメータがない → 1点のみ)
(c) N(μ, Σ) in R^d: 
    - μ: d個のパラメータ
    - Σ: d×d対称正定値行列 → d(d+1)/2 個のパラメータ
    - 合計: d + d(d+1)/2 = d(d+3)/2 次元
    - 例: d=2 なら 2(2+3)/2 = 5次元
""")
```

```python
# Q2の解答
print("""
Q2 解答:
座標変換: (μ, σ) → (μ, τ) where τ = 1/σ²

新座標を旧座標で表す:
  μ' = μ
  τ = σ⁻²

ヤコビ行列 J = ∂(μ', τ)/∂(μ, σ):

J = [[∂μ'/∂μ,  ∂μ'/∂σ ],
     [∂τ/∂μ,   ∂τ/∂σ  ]]
  
  = [[1,       0      ],
     [0,       -2/σ³  ]]

det(J) = -2/σ³ ≠ 0 (σ > 0 なので)
→ 座標変換は正則（逆変換可能）
""")
```

```python
# Q3の解答
print("""
Q3 解答:
カルマンフィルタの経験から:

1. 観測精度が高い（σ_obs が小さい）とき:
   - カルマンゲイン K が大きくなる
   - 状態の更新量が大きい
   - つまり、高精度の観測は「影響力が大きい」

2. 情報幾何的解釈:
   - σが小さい分布は「鋭い」（peakが高い）
   - 微小なパラメータ変化でも、分布の形が大きく変わる
   - これがFisher情報が大きいことの意味
   - 「情報的に遠い」= 小さな座標変化で大きな情報変化

3. 具体例:
   - N(0, 0.1²) と N(0.01, 0.1²) は見た目が大きく異なる
   - N(0, 10²) と N(0.01, 10²) はほぼ同じに見える
   - 同じ Δμ = 0.01 でも、σが小さいほど「大きな移動」
""")
```
