## 情報幾何の核心的アイデア（プレビュー）

情報幾何学の核心は、**確率分布の集合を幾何学的対象（多様体）として捉える**ことです。

### 直感的な対応関係

```
確率・統計の世界              情報幾何の世界
─────────────────────────────────────────────
確率分布 p(x|θ)        ←→   多様体上の点
パラメータ θ           ←→   局所座標
パラメータの変化 dθ     ←→   接ベクトル
Fisher情報行列 I(θ)    ←→   リーマン計量
KLダイバージェンス      ←→   ダイバージェンス関数
最尤推定               ←→   射影（m-射影）
ベイズ更新             ←→   測地線上の移動
```

## なぜ幾何学的視点が有用か？

### 1. 座標に依存しない理解
- 同じ分布族を (μ, σ) でも (μ, σ²) でも (μ, 1/σ²) でも表現できる
- 幾何学的量（曲率、測地線など）は座標の取り方に依存しない

### 2. 最適化の効率化（自然勾配）
- 通常の勾配降下は「座標依存」
- 自然勾配はFisher計量を考慮した「真の最急降下方向」
- ニューラルネットの学習で実際に使われている

### 3. 機械学習アルゴリズムの統一的理解
- EMアルゴリズム → 交互射影
- 変分推論 → 部分多様体への射影
- ベイズ推定 → 測地線と射影の組み合わせ
