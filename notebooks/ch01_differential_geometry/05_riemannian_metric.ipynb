# §1.5 リーマン計量

多様体上で「距離」や「角度」を測る道具としてのリーマン計量を、Fisher情報行列との関係から理解します。

---
## 1. 概要

### この節で学ぶこと

- **リーマン計量**とは何か（接空間上の内積）
- **Fisher情報行列がリーマン計量になる**という重要な事実
- KLダイバージェンスの2次近似との関係
- 計量による距離の計算方法

### 既知概念との対応

| 情報幾何の概念 | 既知概念 |
|--------------|--------|
| リーマン計量テンソル | Fisher情報行列 |
| リーマン距離の2乗 | KLダイバージェンス（2次近似）|
| 計量の定義 | スコア関数の共分散 |

---
## 2. 直感的理解

### リーマン計量とは「曲がった空間での定規」

**身近な例：地球上の距離**
- 地球上では、ユークリッド距離では正しい距離が測れない
- 緯度・経度で表された2点間の「本当の距離」は測地線に沿って測る
- リーマン計量は、この「各点での距離の測り方」を指定する

### 統計的多様体での計量

**Fisher情報行列が計量になる理由**
- パラメータ空間では、座標の差だけでは「分布の違い」を正しく測れない
- $\sigma$ が小さい領域では、同じ $\Delta\mu$ でも分布が大きく変わる
- **Fisher情報行列は、この「情報的な距離」を正しく測る計量**

### 核心的な等式

KLダイバージェンスの2次近似がリーマン距離の2乗になる：

$$D_{\text{KL}}(p_\theta \| p_{\theta+d\theta}) \approx \frac{1}{2} d\theta^\top I(\theta) d\theta$$

**これが「情報幾何」の名前の由来**：情報理論的な量（KLダイバージェンス）が幾何学的な量（距離）になる

---
## 3. 数学的定義

### 定義：リーマン計量

多様体 $M$ 上の**リーマン計量** $g$ とは、各点 $p \in M$ の接空間 $T_pM$ 上に定義された**正定値対称双線形形式**（内積）の滑らかな族：

$$g_p: T_pM \times T_pM \to \mathbb{R}$$

座標 $(\theta^1, \ldots, \theta^n)$ での成分表示：
$$g_{ij}(\theta) = g\left(\frac{\partial}{\partial\theta^i}, \frac{\partial}{\partial\theta^j}\right)$$

### Fisher計量

統計的多様体において、**Fisher情報行列**がリーマン計量の役割を果たす：

$$I_{ij}(\theta) = E\left[\frac{\partial \log p(x|\theta)}{\partial \theta^i} \cdot \frac{\partial \log p(x|\theta)}{\partial \theta^j}\right]$$

### Fisher計量による距離（接ベクトルの長さ）

接ベクトル $v = (v^1, v^2, \ldots, v^n)$ のFisher長さ：
$$||v||^2_{\text{Fisher}} = v^\top I(\theta) v = \sum_{i,j} I_{ij}(\theta) v^i v^j$$

### 正規分布のFisher情報行列

$$I(\mu, \sigma) = \begin{pmatrix} 1/\sigma^2 & 0 \\ 0 & 2/\sigma^2 \end{pmatrix}$$

各成分の意味：
- $I_{\mu\mu} = 1/\sigma^2$: μに関する情報量（σが小さいほど大）
- $I_{\sigma\sigma} = 2/\sigma^2$: σに関する情報量
- $I_{\mu\sigma} = 0$: μとσは統計的に独立

### KLダイバージェンスとFisher計量の関係

$$D_{\text{KL}}(p_\theta \| p_{\theta+d\theta}) = \frac{1}{2} d\theta^\top I(\theta) d\theta + O(||d\theta||^3)$$

この等式により、Fisher情報行列がリーマン計量として自然であることが示される。

## 4. 可視化

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
````

````python
def gaussian_pdf(x, mu, sigma):
    """正規分布の確率密度関数"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

def compute_fisher_gaussian(mu, sigma):
    """
    正規分布 N(μ, σ²) のFisher情報行列
    I = [[1/σ², 0], [0, 2/σ²]]
    """
    return np.array([[1/sigma**2, 0], [0, 2/sigma**2]])

def kl_divergence_gaussian(mu1, sigma1, mu2, sigma2):
    """D_KL(N(μ1,σ1²) || N(μ2,σ2²))"""
    return np.log(sigma2/sigma1) + (sigma1**2 + (mu1-mu2)**2)/(2*sigma2**2) - 0.5
````

### 4.1 Fisher情報行列の成分

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sigma_range = np.linspace(0.3, 3, 100)

# I_μμ = 1/σ²
ax1 = axes[0]
I_mu_mu = 1 / sigma_range**2
ax1.plot(sigma_range, I_mu_mu, 'r-', linewidth=2)
ax1.set_xlabel('σ', fontsize=12)
ax1.set_ylabel('I_μμ = 1/σ²', fontsize=12)
ax1.set_title('Fisher information for μ\nSmaller σ → more information')
ax1.grid(True, alpha=0.3)

# I_σσ = 2/σ²
ax2 = axes[1]
I_sigma_sigma = 2 / sigma_range**2
ax2.plot(sigma_range, I_sigma_sigma, 'b-', linewidth=2)
ax2.set_xlabel('σ', fontsize=12)
ax2.set_ylabel('I_σσ = 2/σ²', fontsize=12)
ax2.set_title('Fisher information for σ')
ax2.grid(True, alpha=0.3)

# 情報行列の楕円表示
ax3 = axes[2]
theta = np.linspace(0, 2*np.pi, 100)
points = [(0, 0.5, 'red'), (0, 1.0, 'blue'), (0, 2.0, 'green')]

for mu, sigma, color in points:
    scale = 0.15
    ellipse_x = sigma * np.cos(theta) * scale + mu
    ellipse_y = sigma/np.sqrt(2) * np.sin(theta) * scale + sigma
    ax3.plot(ellipse_x, ellipse_y, color=color, linewidth=2, label=f'σ={sigma}')
    ax3.plot(mu, sigma, 'o', color=color, markersize=8)

ax3.set_xlabel('μ', fontsize=12)
ax3.set_ylabel('σ', fontsize=12)
ax3.set_title('Fisher metric ellipses\nSmaller σ → smaller ellipse')
ax3.legend()
ax3.set_xlim(-0.5, 0.5)
ax3.set_ylim(0, 2.5)
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.2 ユークリッド計量 vs Fisher計量

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

mu0, sigma0 = 0, 1
theta = np.linspace(0, 2*np.pi, 100)

# 左図：ユークリッド計量 vs Fisher計量での「単位円」
ax1 = axes[0]

# 単位円（ユークリッド）
x_euclid = np.cos(theta) * 0.5
y_euclid = np.sin(theta) * 0.5
ax1.plot(x_euclid + mu0, y_euclid + sigma0, 'b--', linewidth=2, label='Euclidean unit circle')

# Fisher計量による「単位円」（楕円になる）
x_fisher = sigma0 * np.cos(theta) * 0.5
y_fisher = sigma0 / np.sqrt(2) * np.sin(theta) * 0.5
ax1.plot(x_fisher + mu0, y_fisher + sigma0, 'r-', linewidth=2, label='Fisher unit "circle"')

ax1.plot(mu0, sigma0, 'ko', markersize=10)
ax1.text(mu0 - 0.1, sigma0 - 0.15, 'p', fontsize=12, fontweight='bold')

ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('Euclidean vs Fisher metric\n"Unit ball" shapes differ', fontsize=11)
ax1.legend()
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)

# 右図：異なる点での計量楕円
ax2 = axes[1]

points = [
    (0, 0.5, 'red'),
    (1, 1.0, 'blue'),
    (-1, 1.5, 'green'),
    (0, 2.0, 'purple'),
]

for mu, sigma, color in points:
    x = sigma * np.cos(theta) * 0.2
    y = sigma / np.sqrt(2) * np.sin(theta) * 0.2
    ax2.plot(x + mu, y + sigma, color=color, linewidth=1.5)
    ax2.plot(mu, sigma, 'o', color=color, markersize=8)

ax2.set_xlabel('μ', fontsize=12)
ax2.set_ylabel('σ', fontsize=12)
ax2.set_title('Fisher metric ellipses at different points\nSmaller σ → smaller ellipse (feels "farther")', fontsize=11)
ax2.set_xlim(-2, 2)
ax2.set_ylim(0, 3)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.3 KLダイバージェンスと2次近似の比較

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

mu0, sigma0 = 0, 1

mu_range = np.linspace(-1, 1, 100)
sigma_range = np.linspace(0.5, 1.5, 100)
MU, SIGMA = np.meshgrid(mu_range, sigma_range)

# KLダイバージェンス D_KL(N(μ₀,σ₀²) || N(μ,σ²))
KL = np.log(SIGMA/sigma0) + (sigma0**2 + (mu0-MU)**2)/(2*SIGMA**2) - 0.5

# 2次近似 (1/2)(dμ²/σ₀² + 2dσ²/σ₀²)
dMU = MU - mu0
dSIGMA = SIGMA - sigma0
QUAD_APPROX = 0.5 * (dMU**2 / sigma0**2 + 2 * dSIGMA**2 / sigma0**2)

levels = [0.01, 0.05, 0.1, 0.2, 0.5]

# 左図：真のKL等高線
ax1 = axes[0]
cs1 = ax1.contour(MU, SIGMA, KL, levels=levels, colors='blue')
ax1.clabel(cs1, inline=True, fontsize=8)
ax1.plot(mu0, sigma0, 'r*', markersize=15, label='Reference point')
ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('True KL divergence contours', fontsize=11)
ax1.legend()
ax1.grid(True, alpha=0.3)

# 右図：2次近似（Fisher計量による）
ax2 = axes[1]
cs2 = ax2.contour(MU, SIGMA, QUAD_APPROX, levels=levels, colors='red')
ax2.clabel(cs2, inline=True, fontsize=8)
ax2.plot(mu0, sigma0, 'r*', markersize=15, label='Reference point')
ax2.set_xlabel('μ', fontsize=12)
ax2.set_ylabel('σ', fontsize=12)
ax2.set_title('Quadratic approximation (Fisher metric)\n(1/2)dθᵀ I dθ', fontsize=11)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.4 同じ座標距離でもFisher距離は異なる

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左図：同じ座標距離でもFisher距離は異なる
ax1 = axes[0]

# 経路1: σが小さい領域 (0.5 → 0.6)
mu1, sigma1_start, sigma1_end = 0, 0.5, 0.6
ax1.plot([mu1, mu1], [sigma1_start, sigma1_end], 'r-', linewidth=3,
         label=f'Path 1: σ={sigma1_start}→{sigma1_end}')
ax1.plot(mu1, sigma1_start, 'ro', markersize=10)
ax1.plot(mu1, sigma1_end, 'rs', markersize=10)

# 経路2: σが大きい領域 (2.0 → 2.1)
mu2, sigma2_start, sigma2_end = 1, 2.0, 2.1
ax1.plot([mu2, mu2], [sigma2_start, sigma2_end], 'b-', linewidth=3,
         label=f'Path 2: σ={sigma2_start}→{sigma2_end}')
ax1.plot(mu2, sigma2_start, 'bo', markersize=10)
ax1.plot(mu2, sigma2_end, 'bs', markersize=10)

# Fisher距離の計算
d_sigma = 0.1
fisher_dist_1 = np.sqrt(2) * d_sigma / sigma1_start
fisher_dist_2 = np.sqrt(2) * d_sigma / sigma2_start

ax1.text(mu1 + 0.1, (sigma1_start + sigma1_end)/2,
         f'Fisher dist≈{fisher_dist_1:.3f}', fontsize=10, color='red')
ax1.text(mu2 + 0.1, (sigma2_start + sigma2_end)/2,
         f'Fisher dist≈{fisher_dist_2:.3f}', fontsize=10, color='blue')

ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('Same coordinate distance, different Fisher distance\nSmall σ region feels "farther"', fontsize=11)
ax1.legend()
ax1.set_xlim(-0.5, 2)
ax1.set_ylim(0, 3)
ax1.grid(True, alpha=0.3)

# 右図：対応する分布の変化
ax2 = axes[1]
x = np.linspace(-4, 6, 200)

# 経路1の分布変化
y1_start = gaussian_pdf(x, mu1, sigma1_start)
y1_end = gaussian_pdf(x, mu1, sigma1_end)
ax2.plot(x, y1_start, 'r-', linewidth=2, label=f'σ={sigma1_start}')
ax2.plot(x, y1_end, 'r--', linewidth=2, label=f'σ={sigma1_end}')

# 経路2の分布変化
y2_start = gaussian_pdf(x, mu2, sigma2_start)
y2_end = gaussian_pdf(x, mu2, sigma2_end)
ax2.plot(x, y2_start, 'b-', linewidth=2, label=f'σ={sigma2_start}')
ax2.plot(x, y2_end, 'b--', linewidth=2, label=f'σ={sigma2_end}')

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('p(x)', fontsize=12)
ax2.set_title('Distribution changes\nNarrow dist (red) shows larger visual change', fontsize=11)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"""
【計算結果】
- Path 1 (σ=0.5→0.6): Fisher distance ≈ {fisher_dist_1:.4f}
- Path 2 (σ=2.0→2.1): Fisher distance ≈ {fisher_dist_2:.4f}
- Ratio: {fisher_dist_1/fisher_dist_2:.2f}x
""")
````

## 5. 具体例

### 例1：正規分布のFisher情報行列

````python
def compute_fisher_gaussian_verbose(mu, sigma):
    """
    正規分布のFisher情報行列を計算（導出過程付き）
    """
    print(f"正規分布 N({mu}, {sigma}²) のFisher情報行列:")
    print("="*50)
    print("")
    print("対数尤度: log p(x|μ,σ) = -log(σ√(2π)) - (x-μ)²/(2σ²)")
    print("")
    print("スコア関数:")
    print("  ∂log p/∂μ = (x-μ)/σ²")
    print("  ∂log p/∂σ = -1/σ + (x-μ)²/σ³")
    print("")
    print("Fisher情報行列の成分:")
    print(f"  I_μμ = E[(∂log p/∂μ)²] = E[(x-μ)²]/σ⁴ = σ²/σ⁴ = 1/σ² = {1/sigma**2:.4f}")
    print(f"  I_σσ = E[(∂log p/∂σ)²] = 2/σ² = {2/sigma**2:.4f}")
    print(f"  I_μσ = E[(∂log p/∂μ)(∂log p/∂σ)] = 0  (独立)")
    print("")
    
    I = compute_fisher_gaussian(mu, sigma)
    print("Fisher情報行列:")
    print(f"  I = [[{I[0,0]:.4f}, {I[0,1]:.4f}],")
    print(f"       [{I[1,0]:.4f}, {I[1,1]:.4f}]]")
    
    return I

# 具体例
I = compute_fisher_gaussian_verbose(0, 1)
````

### 例2：接ベクトルのFisher長さ

````python
def fisher_length(v, mu, sigma):
    """
    接ベクトル v = (v_μ, v_σ) のFisher長さを計算
    """
    I = compute_fisher_gaussian(mu, sigma)
    v_arr = np.array(v)
    length_sq = v_arr @ I @ v_arr
    return np.sqrt(length_sq)

# 具体例
mu, sigma = 0, 1
test_vectors = [
    (1, 0),    # μ方向
    (0, 1),    # σ方向
    (1, 1),    # 斜め方向
    (0.5, 0.5),
]

print(f"点 (μ, σ) = ({mu}, {sigma}) での接ベクトルのFisher長さ:")
print("-" * 50)
I = compute_fisher_gaussian(mu, sigma)
print(f"Fisher情報行列: I = [[{I[0,0]:.2f}, {I[0,1]:.2f}], [{I[1,0]:.2f}, {I[1,1]:.2f}]]")
print("-" * 50)

for v in test_vectors:
    length = fisher_length(v, mu, sigma)
    print(f"v = {v}: ||v||_Fisher = {length:.4f}")
````

### 例3：ポアソン分布のFisher情報

````python
def fisher_poisson(lam):
    """
    ポアソン分布 Poisson(λ) のFisher情報
    I(λ) = 1/λ
    """
    return 1 / lam

# 可視化
lam_range = np.linspace(0.1, 5, 100)
fisher_values = fisher_poisson(lam_range)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(lam_range, fisher_values, 'b-', linewidth=2)
ax.set_xlabel('λ', fontsize=12)
ax.set_ylabel('I(λ) = 1/λ', fontsize=12)
ax.set_title('Poisson distribution Fisher information\nSmaller λ → more information', fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("ポアソン分布のFisher情報 I(λ) = 1/λ の意味:")
print("- λが小さいほどFisher情報が大きい")
print("- λが小さい領域では、パラメータの微小変化が分布に大きく影響")
print("- Cramér-Raoの下界: Var(λ̂) ≥ λ/n  (MLEでは等号成立)")
````

## 6. 他の概念との関係

### 6.1 カルマンフィルタとの接点

カルマンフィルタの更新において：
- 観測精度が高い（$\sigma_{\text{obs}}$ が小さい）→ Fisher情報が大きい
- → カルマンゲイン $K$ が大きい
- → 多様体上で「遠く」まで移動できる

**これがFisher計量の「距離が大きく感じる」ことの実践的意味**

````python
# 観測精度とFisher距離の関係
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 事前分布
mu_pred, sigma_pred = 0, 1.5
mu_obs = 1.0

# 異なる観測精度
sigma_obs_values = [0.3, 0.8, 1.5]
colors = ['red', 'blue', 'green']

ax1 = axes[0]
ax1.plot(mu_pred, sigma_pred, 'ko', markersize=15, label='Prior')

for sigma_obs, color in zip(sigma_obs_values, colors):
    K = sigma_pred**2 / (sigma_pred**2 + sigma_obs**2)
    mu_post = mu_pred + K * (mu_obs - mu_pred)
    sigma_post = np.sqrt((1 - K) * sigma_pred**2)
    
    ax1.plot(mu_post, sigma_post, 'o', color=color, markersize=12, 
             label=f'σ_obs={sigma_obs}, K={K:.2f}')
    ax1.annotate('', xy=(mu_post, sigma_post), xytext=(mu_pred, sigma_pred),
                 arrowprops=dict(arrowstyle='->', color=color, lw=1.5))

ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('High observation precision →\nLarge movement on manifold', fontsize=11)
ax1.legend(fontsize=9)
ax1.set_xlim(-0.5, 1.5)
ax1.set_ylim(0, 2)
ax1.grid(True, alpha=0.3)

# 右図：カルマンゲインと観測精度の関係
ax2 = axes[1]
sigma_obs_range = np.linspace(0.1, 3, 100)
K_values = sigma_pred**2 / (sigma_pred**2 + sigma_obs_range**2)

ax2.plot(sigma_obs_range, K_values, 'b-', linewidth=2)
ax2.set_xlabel('σ_obs (observation noise)', fontsize=12)
ax2.set_ylabel('Kalman gain K', fontsize=12)
ax2.set_title('Kalman gain vs observation precision\nSmaller σ_obs → larger K → larger update', fontsize=11)
ax2.grid(True, alpha=0.3)

for sigma_obs, color in zip(sigma_obs_values, colors):
    K = sigma_pred**2 / (sigma_pred**2 + sigma_obs**2)
    ax2.plot(sigma_obs, K, 'o', color=color, markersize=10)

plt.tight_layout()
plt.show()
````

### 6.2 自然勾配法との接点

通常の勾配降下法では、ユークリッド計量での勾配を使う：
$$\theta_{t+1} = \theta_t - \eta \nabla L(\theta_t)$$

**自然勾配法**では、Fisher計量での勾配を使う：
$$\theta_{t+1} = \theta_t - \eta I(\theta_t)^{-1} \nabla L(\theta_t)$$

- パラメータの取り方に依存しない更新
- 統計的多様体の「本質的な」方向に進む

### 6.3 後続の概念への橋渡し

| 概念 | リーマン計量との関係 |
|------|---------------------|
| 測地線 | 計量による「最短経路」 |
| 接続 | 接ベクトルの平行移動を定義 |
| 曲率 | 多様体の「曲がり具合」 |

## 7. 演習問題

### Q1. Fisher情報の計算

ベルヌーイ分布 $\text{Ber}(p)$ のFisher情報 $I(p)$ を計算せよ。

<details>
<summary><b>解答を見る</b></summary>

ベルヌーイ分布: $P(x|p) = p^x (1-p)^{1-x}$

スコア関数（前節で計算済み）:
$$\frac{\partial \log P}{\partial p} = \frac{x - p}{p(1-p)}$$

Fisher情報:
$$I(p) = E\left[\left(\frac{\partial \log P}{\partial p}\right)^2\right] = E\left[\frac{(x-p)^2}{p^2(1-p)^2}\right]$$

$E[(x-p)^2] = \text{Var}(x) = p(1-p)$ より:

$$I(p) = \frac{p(1-p)}{p^2(1-p)^2} = \frac{1}{p(1-p)}$$

**答: $I(p) = \frac{1}{p(1-p)}$**

（$p = 0.5$ で最小、端点に近づくほど大きくなる）

</details>

### Q2. 計量による長さ

正規分布の多様体で、点 $(\mu, \sigma) = (0, 1)$ における接ベクトル $v = (1, 1)$ のFisher長さを計算せよ。

<details>
<summary><b>解答を見る</b></summary>

点 $(\mu, \sigma) = (0, 1)$ でのFisher情報行列:
$$I = \begin{pmatrix} 1/\sigma^2 & 0 \\ 0 & 2/\sigma^2 \end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & 2 \end{pmatrix}$$

接ベクトル $v = (1, 1)$ のFisher長さの2乗:
$$||v||^2 = v^\top I v = (1, 1) \begin{pmatrix} 1 & 0 \\ 0 & 2 \end{pmatrix} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = 1 \cdot 1 + 2 \cdot 1 = 3$$

**答: $||v||_{\text{Fisher}} = \sqrt{3} \approx 1.732$**

</details>

````python
# Q2の計算確認
mu, sigma = 0, 1
v = np.array([1, 1])
I = compute_fisher_gaussian(mu, sigma)

fisher_length_sq = v @ I @ v
fisher_length = np.sqrt(fisher_length_sq)

print(f"Q2 解答確認:")
print(f"Fisher情報行列 I = {I}")
print(f"接ベクトル v = {v}")
print(f"||v||² = v'Iv = {fisher_length_sq}")
print(f"||v|| = √{fisher_length_sq} = {fisher_length:.4f}")
````

### Q3. 距離の比較

$(0, 1) \to (0.1, 1)$ と $(0, 2) \to (0.1, 2)$ の移動で、どちらがFisher距離が大きいか？

<details>
<summary><b>解答を見る</b></summary>

両方とも $d\mu = 0.1$, $d\sigma = 0$ の移動。

**移動1: $(0, 1) \to (0.1, 1)$**
$$\text{Fisher距離} = \sqrt{d\mu^2 / \sigma^2} = \sqrt{0.1^2 / 1^2} = 0.1$$

**移動2: $(0, 2) \to (0.1, 2)$**
$$\text{Fisher距離} = \sqrt{d\mu^2 / \sigma^2} = \sqrt{0.1^2 / 2^2} = 0.05$$

**答: 移動1の方がFisher距離が大きい（2倍）**

直感: $\sigma$ が小さい領域では、同じ $\mu$ の変化でも「情報的に」大きな変化

</details>

````python
# Q3の計算確認
d_mu, d_sigma = 0.1, 0

# 移動1: σ=1
sigma1 = 1
dist1 = np.sqrt(d_mu**2 / sigma1**2)

# 移動2: σ=2
sigma2 = 2
dist2 = np.sqrt(d_mu**2 / sigma2**2)

print(f"Q3 解答確認:")
print(f"移動1 (σ=1): Fisher距離 = {dist1:.4f}")
print(f"移動2 (σ=2): Fisher距離 = {dist2:.4f}")
print(f"比: {dist1/dist2:.1f}倍")
````
