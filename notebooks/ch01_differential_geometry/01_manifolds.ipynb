# §1.1 微分可能多様体

情報幾何の土台となる「多様体」の概念を、正規分布のパラメータ空間を例に理解します。

---
## 1. 概要

### この節で学ぶこと

- **多様体**とは何か（直感的・数学的定義）
- 確率分布のパラメータ空間が多様体を形成すること
- **局所座標**と**座標変換**の概念
- 統計的多様体における「曲がり」の意味

### 既知概念との対応

| 情報幾何の概念 | 既知概念 |
|--------------|--------|
| 多様体 | パラメータ空間 $(\mu, \sigma)$ |
| 多様体上の点 | パラメータの組、つまり一つの確率分布 |
| 局所座標 | パラメータ化の方法 |
| 座標変換 | 異なるパラメータ化間の変換 |

---
## 2. 直感的理解

### 多様体とは「局所的に平らに見える曲がった空間」

**身近な例：地球の表面**
- 地球は3次元空間に埋め込まれた2次元の曲面
- 小さい領域を見れば平面のように見える（地図が作れる）
- しかし全体としては曲がっている（地球儀でないと正確に表せない）

### 統計的多様体の直感

**正規分布のパラメータ空間を考える**
- 正規分布 $N(\mu, \sigma^2)$ は2つのパラメータで特徴づけられる
- パラメータの組 $(\mu, \sigma)$ の空間は2次元の多様体
- **各点が一つの確率分布に対応**
- $\sigma > 0$ の制約があるため、上半平面（境界を含まない）

### なぜ「多様体」という言葉を使うのか

- パラメータ空間は一見平面に見えるが...
- **Fisher計量を入れると曲がった空間になる**
- $\sigma$ が小さい領域は「Fisher的に遠い」
- この「曲がり」を正確に扱うために多様体の概念が必要

---
## 3. 数学的定義

### 定義：$n$次元微分可能多様体

集合 $M$ が**$n$次元微分可能多様体**であるとは、以下を満たすことをいう：

1. **局所座標系の存在**: 各点 $p \in M$ の近傍 $U$ に対して、$\mathbb{R}^n$ の開集合 $V$ への同相写像（**座標写像**）$\varphi: U \to V$ が存在

2. **座標変換の滑らかさ**: 二つの座標系 $(U_\alpha, \varphi_\alpha)$, $(U_\beta, \varphi_\beta)$ の重なり $U_\alpha \cap U_\beta \neq \emptyset$ において、
   $$\varphi_\beta \circ \varphi_\alpha^{-1}: \varphi_\alpha(U_\alpha \cap U_\beta) \to \varphi_\beta(U_\alpha \cap U_\beta)$$
   が $C^\infty$級（無限回微分可能）

### 正規分布の多様体

正規分布族 $\mathcal{M} = \{N(\mu, \sigma^2) : \mu \in \mathbb{R}, \sigma > 0\}$ は**2次元多様体**：

- 各点 $p \in \mathcal{M}$ は確率分布 $N(\mu, \sigma^2)$
- 座標: $(\mu, \sigma) \in \mathbb{R} \times \mathbb{R}_+$
- 多様体の次元 = パラメータの数 = 2

### 座標変換の例

同じ多様体を異なる座標系で表現できる：

| 座標系 | パラメータ | 用途 |
|-------|----------|------|
| 標準座標 | $(\mu, \sigma)$ | 自然な座標 |
| 分散座標 | $(\mu, \sigma^2)$ | 分散を直接使う |
| 精度座標 | $(\mu, \tau = 1/\sigma^2)$ | ベイズ推定で頻出 |

座標変換 $(\mu, \sigma) \to (\mu, \tau)$ のヤコビ行列：

$$J = \frac{\partial(\mu, \tau)}{\partial(\mu, \sigma)} = \begin{pmatrix} 1 & 0 \\ 0 & -2/\sigma^3 \end{pmatrix}$$

$\det(J) = -2/\sigma^3 \neq 0$ より、座標変換は正則（逆変換可能）

## 4. 可視化

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
````

````python
def gaussian_pdf(x, mu, sigma):
    """正規分布の確率密度関数"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
````

### 4.1 パラメータ空間 = 多様体

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左図：パラメータ空間（多様体）
ax1 = axes[0]
mu_range = np.linspace(-3, 3, 20)
sigma_range = np.linspace(0.1, 3, 20)

# グリッド点をプロット
for mu in mu_range[::2]:
    for sigma in sigma_range[::2]:
        ax1.plot(mu, sigma, 'b.', markersize=3, alpha=0.5)

# いくつかの特定の分布を強調
special_points = [
    (0, 1, 'r', 'Standard normal N(0,1)'),
    (1, 0.5, 'g', 'Narrow N(1, 0.5²)'),
    (-1, 2, 'm', 'Wide N(-1, 2²)'),
]

for mu, sigma, color, label in special_points:
    ax1.plot(mu, sigma, 'o', color=color, markersize=12, label=label)

ax1.set_xlabel('μ (mean)', fontsize=12)
ax1.set_ylabel('σ (std dev)', fontsize=12)
ax1.set_title('Parameter space = Statistical manifold\nEach point = one probability distribution', fontsize=11)
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3, label='σ=0 boundary (not included)')
ax1.legend(loc='upper right', fontsize=9)
ax1.set_xlim(-4, 4)
ax1.set_ylim(-0.5, 4)
ax1.grid(True, alpha=0.3)

# 右図：対応する確率密度関数
ax2 = axes[1]
x = np.linspace(-6, 6, 200)

for mu, sigma, color, label in special_points:
    y = gaussian_pdf(x, mu, sigma)
    ax2.plot(x, y, color=color, linewidth=2, label=label)

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('p(x)', fontsize=12)
ax2.set_title('Corresponding probability density functions', fontsize=11)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.2 異なる座標系での表現

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 同じ分布族を異なる座標系で表示
mu_vals = np.linspace(-2, 2, 10)
sigma_vals = np.linspace(0.5, 2, 10)

# 座標系1: (μ, σ)
ax1 = axes[0]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax1.plot(mu, sigma, 'b.', markersize=4)
ax1.set_xlabel('μ')
ax1.set_ylabel('σ')
ax1.set_title('Coordinate 1: (μ, σ)')
ax1.grid(True, alpha=0.3)

# 座標系2: (μ, σ²)
ax2 = axes[1]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax2.plot(mu, sigma**2, 'g.', markersize=4)
ax2.set_xlabel('μ')
ax2.set_ylabel('σ² (variance)')
ax2.set_title('Coordinate 2: (μ, σ²)')
ax2.grid(True, alpha=0.3)

# 座標系3: (μ, 1/σ²) - 精度パラメータ
ax3 = axes[2]
for mu in mu_vals:
    for sigma in sigma_vals:
        ax3.plot(mu, 1/sigma**2, 'r.', markersize=4)
ax3.set_xlabel('μ')
ax3.set_ylabel('τ = 1/σ² (precision)')
ax3.set_title('Coordinate 3: (μ, τ)\nCommon in Bayesian inference')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.3 多様体の「曲がり」と球面との比較

````python
fig = plt.figure(figsize=(14, 5))

# 左図：球面（2次元多様体の典型例）
ax1 = fig.add_subplot(121, projection='3d')
u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))
ax1.plot_surface(x, y, z, alpha=0.6, cmap='viridis')
ax1.set_title('Sphere: 2D manifold in 3D space\nLocally looks flat', fontsize=11)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('z')

# 右図：正規分布の多様体を「曲がった空間」として描画
ax2 = fig.add_subplot(122)

# ユークリッド的なグリッド
mu_range = np.linspace(-2, 2, 11)
sigma_range = np.linspace(0.3, 2, 11)

# 通常のグリッド（薄い線）
for mu in mu_range:
    ax2.plot([mu, mu], [0.3, 2], 'b-', alpha=0.2)
for sigma in sigma_range:
    ax2.plot([-2, 2], [sigma, sigma], 'b-', alpha=0.2)

# Fisher計量による「等距離線」（概念的）
for i, sigma in enumerate([0.5, 1.0, 1.5]):
    mu_dense = np.linspace(-2, 2, int(20/sigma))
    for mu in mu_dense:
        ax2.plot(mu, sigma, 'r.', markersize=3)

ax2.annotate('Small σ region:\n"Fisher-far"',
             xy=(0, 0.5), xytext=(1.5, 1.2),
             arrowprops=dict(arrowstyle='->', color='red'),
             fontsize=10, color='red')

ax2.set_xlabel('μ', fontsize=12)
ax2.set_ylabel('σ', fontsize=12)
ax2.set_title('Gaussian manifold\n"Curved" with Fisher metric', fontsize=11)
ax2.set_xlim(-2.5, 2.5)
ax2.set_ylim(0, 2.5)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 5. 具体例

### 例1：正規分布の多様体

$$\mathcal{M}_{\text{Gauss}} = \{N(\mu, \sigma^2) : \mu \in \mathbb{R}, \sigma > 0\}$$

- 次元: 2
- パラメータ空間: $\mathbb{R} \times \mathbb{R}_+$ （上半平面）

````python
# 正規分布多様体上の曲線（パラメータの連続変化）
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# パラメータ空間での曲線
t = np.linspace(0, 2*np.pi, 100)
mu_curve = np.sin(t)
sigma_curve = 1 + 0.5 * np.cos(t)

ax1 = axes[0]
ax1.plot(mu_curve, sigma_curve, 'b-', linewidth=2)
ax1.plot(mu_curve[0], sigma_curve[0], 'go', markersize=12, label='Start')
ax1.plot(mu_curve[-1], sigma_curve[-1], 'rs', markersize=12, label='End')
ax1.set_xlabel('μ')
ax1.set_ylabel('σ')
ax1.set_title('A curve on the Gaussian manifold\n(continuous change of parameters)')
ax1.legend()
ax1.grid(True, alpha=0.3)
ax1.set_xlim(-1.5, 1.5)
ax1.set_ylim(0, 2)

# 対応する分布の変化（アニメーション的に数点を表示）
ax2 = axes[1]
x = np.linspace(-4, 4, 200)
colors = plt.cm.viridis(np.linspace(0, 1, 5))

for i, (idx, c) in enumerate(zip([0, 25, 50, 75, 99], colors)):
    mu, sigma = mu_curve[idx], sigma_curve[idx]
    y = gaussian_pdf(x, mu, sigma)
    ax2.plot(x, y, color=c, linewidth=2, label=f't={idx/100:.2f}: (μ,σ)=({mu:.2f},{sigma:.2f})')

ax2.set_xlabel('x')
ax2.set_ylabel('p(x)')
ax2.set_title('Distributions along the curve')
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 例2：多次元正規分布の多様体の次元

````python
def manifold_dimension_multivariate_gaussian(d):
    """
    d次元多変量正規分布 N(μ, Σ) の多様体の次元を計算
    
    Parameters:
    -----------
    d : int
        データの次元
    
    Returns:
    --------
    int : 多様体の次元
    """
    # μ: d個のパラメータ
    # Σ: d×d対称正定値行列 → d(d+1)/2 個のパラメータ
    return d + d * (d + 1) // 2

print("多次元正規分布 N(μ, Σ) の多様体の次元:")
print("-" * 40)
for d in range(1, 6):
    dim = manifold_dimension_multivariate_gaussian(d)
    print(f"d = {d}: μの次元 = {d}, Σのパラメータ数 = {d*(d+1)//2}, 多様体次元 = {dim}")
````

### 例3：座標変換の計算

````python
def coordinate_transform_jacobian(sigma):
    """
    座標変換 (μ, σ) → (μ, τ=1/σ²) のヤコビ行列
    """
    return np.array([[1, 0], [0, -2/sigma**3]])

# 具体的な計算例
sigma_test = 1.0
J = coordinate_transform_jacobian(sigma_test)
print(f"σ = {sigma_test} でのヤコビ行列:")
print(J)
print(f"\n行列式 det(J) = {np.linalg.det(J):.4f}")
print(f"det(J) ≠ 0 より、座標変換は正則（可逆）")
````

## 6. 他の概念との関係

### 6.1 カルマンフィルタとの接点

カルマンフィルタでの状態更新は**多様体上の移動**として捉えられる：

- 予測分布 $N(\mu_{\text{pred}}, \sigma^2_{\text{pred}})$ → 多様体上の1点
- 観測を得て更新 → 多様体上の別の点へ移動
- この「移動」の方向が**接ベクトル**（次節）

### 6.2 ベイズ推定との接点

ベイズ推定では精度 $\tau = 1/\sigma^2$ を使うことが多い：

- 事前分布と尤度の積が計算しやすい
- 共役事前分布の自然なパラメータ
- これは**座標変換**に対応

### 6.3 後続の概念への橋渡し

| 概念 | 多様体との関係 |
|------|---------------|
| 接ベクトル（§1.2） | 多様体上の点での「方向」 |
| リーマン計量（§1.5） | 多様体上の「距離の測り方」 |
| Fisher情報行列 | 統計的多様体のリーマン計量 |

````python
# カルマンフィルタの更新を多様体上の移動として可視化
fig, ax = plt.subplots(figsize=(10, 6))

# パラメータ
mu_pred, sigma_pred = 0, 2.0   # 事前分布
mu_obs = 1.5                    # 観測値
sigma_obs = 0.8                 # 観測ノイズ

# カルマンゲインと更新
K = sigma_pred**2 / (sigma_pred**2 + sigma_obs**2)
mu_post = mu_pred + K * (mu_obs - mu_pred)
sigma_post = np.sqrt((1 - K) * sigma_pred**2)

# 事前、事後の点
ax.plot(mu_pred, sigma_pred, 'bo', markersize=15, label='Prior')
ax.plot(mu_post, sigma_post, 'go', markersize=15, label='Posterior')

# 更新の矢印
ax.annotate('', xy=(mu_post, sigma_post), xytext=(mu_pred, sigma_pred),
             arrowprops=dict(arrowstyle='->', color='purple', lw=3))
ax.text((mu_pred + mu_post)/2 + 0.1, (sigma_pred + sigma_post)/2 + 0.15,
         'Bayesian update\n(movement on manifold)', fontsize=10, color='purple')

ax.set_xlabel('μ', fontsize=12)
ax.set_ylabel('σ', fontsize=12)
ax.set_title('Kalman filter update as movement on manifold', fontsize=11)
ax.legend(fontsize=11)
ax.set_xlim(-1, 2.5)
ax.set_ylim(0, 3)
ax.grid(True, alpha=0.3)

plt.show()

print(f"Prior: (μ, σ) = ({mu_pred}, {sigma_pred})")
print(f"Posterior: (μ, σ) = ({mu_post:.3f}, {sigma_post:.3f})")
print(f"Kalman gain K = {K:.3f}")
````

## 7. 演習問題

### Q1. 多様体の次元

以下の統計的多様体の次元を答えよ：

(a) 正規分布 $N(\mu, \sigma^2)$（$\mu, \sigma$ 共に自由）

(b) 標準正規分布 $N(0, 1)$

(c) 分散既知の正規分布 $N(\mu, 1)$（$\mu$ のみ自由）

(d) 多次元正規分布 $N(\mu, \Sigma)$ where $\mu \in \mathbb{R}^3$, $\Sigma$ は $3 \times 3$ 正定値行列

<details>
<summary><b>解答を見る</b></summary>

(a) **2次元** - μとσの2つの自由パラメータ

(b) **0次元** - パラメータがない（1点のみからなる「多様体」）

(c) **1次元** - μのみが自由パラメータ

(d) **9次元**
- μ: 3個のパラメータ
- Σ: 3×3対称正定値行列 → 3(3+1)/2 = 6 個のパラメータ
- 合計: 3 + 6 = 9次元

</details>

### Q2. 座標変換

座標系 $(\mu, \sigma)$ から $(\mu, v = \sigma^2)$（分散座標）への座標変換のヤコビ行列を計算せよ。

<details>
<summary><b>解答を見る</b></summary>

座標変換: $(\mu, \sigma) \to (\mu', v)$ where $\mu' = \mu$, $v = \sigma^2$

ヤコビ行列:
$$J = \frac{\partial(\mu', v)}{\partial(\mu, \sigma)} = \begin{pmatrix} \partial\mu'/\partial\mu & \partial\mu'/\partial\sigma \\ \partial v/\partial\mu & \partial v/\partial\sigma \end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & 2\sigma \end{pmatrix}$$

$\det(J) = 2\sigma > 0$ （$\sigma > 0$より）、座標変換は正則。

</details>

### Q3. 直感の確認

なぜ $\sigma$ が小さい領域は「情報的に遠い」と言えるのか、カルマンフィルタの経験から説明せよ。

<details>
<summary><b>解答を見る</b></summary>

**カルマンフィルタの経験から：**

1. 観測精度が高い（$\sigma_{\text{obs}}$ が小さい）とき：
   - カルマンゲイン $K$ が大きくなる
   - 状態の更新量が大きい
   - つまり、高精度の観測は「影響力が大きい」

2. 情報幾何的解釈：
   - $\sigma$が小さい分布は「鋭い」（peakが高い）
   - 微小なパラメータ変化でも、分布の形が大きく変わる
   - これがFisher情報が大きいことの意味
   - 「情報的に遠い」= 小さな座標変化で大きな情報変化

3. 具体例：
   - $N(0, 0.1^2)$ と $N(0.01, 0.1^2)$ は見た目が大きく異なる
   - $N(0, 10^2)$ と $N(0.01, 10^2)$ はほぼ同じに見える
   - 同じ $\Delta\mu = 0.01$ でも、$\sigma$が小さいほど「大きな移動」

</details>
