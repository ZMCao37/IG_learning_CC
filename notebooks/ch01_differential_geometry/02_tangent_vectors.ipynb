# §1.2 接ベクトルと接空間

多様体上の点における「方向」を表す接ベクトルの概念を、パラメータの微小変化として理解します。

---
## 1. 概要

### この節で学ぶこと

- **接ベクトル**とは何か（多様体上の点での「方向」）
- **接空間**の構造（線形空間としての性質）
- **スコア関数**と接ベクトルの関係
- カルマンフィルタの状態更新との接点

### 既知概念との対応

| 情報幾何の概念 | 既知概念 |
|--------------|--------|
| 接ベクトル | パラメータの微小変化 $(d\mu, d\sigma)$ |
| 接空間 | その点で可能な全ての変化方向 |
| スコア関数 | 対数尤度の勾配（まさに接ベクトル的） |

---
## 2. 直感的理解

### 接ベクトルとは「方向」

**身近な例：地図上の移動方向**
- 地球上の一点に立っている人を考える
- 「北へ進む」「東へ進む」といった**方向**を指定できる
- これらの方向が「接ベクトル」に対応

### 統計的多様体での接ベクトル

パラメータ空間の点 $p = (\mu_0, \sigma_0)$ において：
- 接ベクトルはパラメータの微小変化 $(d\mu, d\sigma)$ に対応
- 「μを増やす方向」「σを減らす方向」などが接ベクトル
- **確率分布の変化方向を指定**

### 接空間は線形空間

点 $p$ での接空間 $T_pM$ は：
- 全ての可能な方向の集合
- **線形空間（ベクトル空間）の構造を持つ**
- 接ベクトル同士の加法、スカラー倍が定義できる

---
## 3. 数学的定義

### 定義：接ベクトル（曲線による定義）

多様体 $M$ 上の点 $p$ における**接ベクトル**とは、$p$ を通る滑らかな曲線
$$\gamma: (-\epsilon, \epsilon) \to M, \quad \gamma(0) = p$$
の $t = 0$ での「速度ベクトル」
$$v = \left.\frac{d\gamma}{dt}\right|_{t=0}$$

### 定義：接空間

点 $p$ における**接空間** $T_pM$ は、$p$ での全ての接ベクトルの集合。
$n$次元多様体の接空間は $n$次元線形空間。

### 座標による表現

座標系 $(\theta^1, \theta^2, \ldots, \theta^n)$ に対して：
- 基底: $\left\{\frac{\partial}{\partial\theta^1}, \frac{\partial}{\partial\theta^2}, \ldots, \frac{\partial}{\partial\theta^n}\right\}$
- 任意の接ベクトル: $v = \sum_{i=1}^n v^i \frac{\partial}{\partial\theta^i}$
- 成分 $(v^1, v^2, \ldots, v^n)$ で表現

### 正規分布の場合

座標 $(\mu, \sigma)$ に対して：
- 基底: $\left\{\frac{\partial}{\partial\mu}, \frac{\partial}{\partial\sigma}\right\}$
- 任意の接ベクトル: $v = a\frac{\partial}{\partial\mu} + b\frac{\partial}{\partial\sigma}$
- 成分表示: $(a, b)$

### スコア関数（統計的多様体での接ベクトル表現）

**スコア関数** = 対数尤度のパラメータ微分
$$s_i(x; \theta) = \frac{\partial \log p(x|\theta)}{\partial \theta^i}$$

これは「接ベクトルを確率変数（関数）として表現」したもの。

正規分布の場合：
$$\frac{\partial \log p(x|\mu,\sigma)}{\partial \mu} = \frac{x - \mu}{\sigma^2}$$
$$\frac{\partial \log p(x|\mu,\sigma)}{\partial \sigma} = -\frac{1}{\sigma} + \frac{(x-\mu)^2}{\sigma^3}$$

## 4. 可視化

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
````

````python
def gaussian_pdf(x, mu, sigma):
    """正規分布の確率密度関数"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
````

### 4.1 パラメータ空間での接ベクトル

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左図：パラメータ空間での接ベクトル
ax1 = axes[0]

# 基準点
mu0, sigma0 = 0, 1
ax1.plot(mu0, sigma0, 'ko', markersize=12, label=f'p = ({mu0}, {sigma0})')

# 接ベクトルの例
tangent_vectors = [
    ((1, 0), 'r', '∂/∂μ direction'),
    ((0, 0.5), 'b', '∂/∂σ direction'),
    ((0.7, 0.7), 'g', 'Mixed direction'),
    ((-0.5, 0.3), 'm', 'Another direction'),
]

for (dmu, dsigma), color, label in tangent_vectors:
    ax1.annotate('', xy=(mu0 + dmu, sigma0 + dsigma), xytext=(mu0, sigma0),
                 arrowprops=dict(arrowstyle='->', color=color, lw=2))
    ax1.text(mu0 + dmu * 1.1, sigma0 + dsigma * 1.1, label, fontsize=9, color=color)

ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('Tangent vectors in parameter space\nRepresent "directions" at point p', fontsize=11)
ax1.set_xlim(-1.5, 2)
ax1.set_ylim(0, 2.5)
ax1.grid(True, alpha=0.3)
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3)

# 右図：分布空間での対応する変化
ax2 = axes[1]
x = np.linspace(-4, 4, 200)

# 基準分布
y0 = gaussian_pdf(x, mu0, sigma0)
ax2.plot(x, y0, 'k-', linewidth=2, label=f'p(x|{mu0},{sigma0}) base')

# 各方向への微小変化
epsilon = 0.3
changes = [
    ((mu0 + epsilon, sigma0), 'r', '--', 'μ+ε'),
    ((mu0, sigma0 + epsilon * 0.5), 'b', '--', 'σ+ε'),
    ((mu0 + epsilon * 0.7, sigma0 + epsilon * 0.7), 'g', '--', 'μ,σ+ε'),
]

for (mu, sigma), color, style, label in changes:
    y = gaussian_pdf(x, mu, sigma)
    ax2.plot(x, y, color=color, linestyle=style, linewidth=1.5, label=label)

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('p(x)', fontsize=12)
ax2.set_title('Corresponding distribution changes\nTangent vector indicates direction of change', fontsize=11)
ax2.legend(loc='upper right')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### 4.2 スコア関数の可視化

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

mu0, sigma0 = 0, 1
x = np.linspace(-4, 4, 200)

# 左上：確率密度関数
ax1 = axes[0, 0]
y = gaussian_pdf(x, mu0, sigma0)
ax1.plot(x, y, 'b-', linewidth=2)
ax1.fill_between(x, y, alpha=0.3)
ax1.set_xlabel('x')
ax1.set_ylabel('p(x)')
ax1.set_title(f'Probability density N({mu0}, {sigma0}²)')
ax1.grid(True, alpha=0.3)

# 右上：対数尤度
ax2 = axes[0, 1]
log_p = np.log(y + 1e-10)  # 数値安定性
ax2.plot(x, log_p, 'g-', linewidth=2)
ax2.set_xlabel('x')
ax2.set_ylabel('log p(x)')
ax2.set_title('Log-likelihood log p(x|μ,σ)')
ax2.grid(True, alpha=0.3)

# 左下：μに関するスコア関数
ax3 = axes[1, 0]
score_mu = (x - mu0) / sigma0**2
ax3.plot(x, score_mu, 'r-', linewidth=2)
ax3.axhline(y=0, color='k', linestyle='--', alpha=0.5)
ax3.set_xlabel('x')
ax3.set_ylabel('∂log p/∂μ')
ax3.set_title('Score function (μ direction)\n= (x-μ)/σ²')
ax3.grid(True, alpha=0.3)

# 右下：σに関するスコア関数
ax4 = axes[1, 1]
score_sigma = -1/sigma0 + (x - mu0)**2 / sigma0**3
ax4.plot(x, score_sigma, 'm-', linewidth=2)
ax4.axhline(y=0, color='k', linestyle='--', alpha=0.5)
ax4.set_xlabel('x')
ax4.set_ylabel('∂log p/∂σ')
ax4.set_title('Score function (σ direction)\n= -1/σ + (x-μ)²/σ³')
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 5. 具体例

### 例1：正規分布のスコア関数の計算

````python
def score_function_gaussian(x, mu, sigma):
    """
    正規分布のスコア関数を計算
    
    Returns:
        score_mu: μに関するスコア
        score_sigma: σに関するスコア
    """
    score_mu = (x - mu) / sigma**2
    score_sigma = -1/sigma + (x - mu)**2 / sigma**3
    return score_mu, score_sigma

# 具体的な計算例
mu, sigma = 0, 1
x_samples = np.array([-1, 0, 1, 2])

print("正規分布 N(0, 1) のスコア関数:")
print("-" * 50)
print(f"{'x':>5} | {'score_μ':>10} | {'score_σ':>10}")
print("-" * 50)
for x_val in x_samples:
    s_mu, s_sigma = score_function_gaussian(x_val, mu, sigma)
    print(f"{x_val:>5} | {s_mu:>10.4f} | {s_sigma:>10.4f}")
````

### 例2：スコア関数の期待値がゼロであることの確認

````python
def verify_score_expectation_zero(mu, sigma, n_samples=100000):
    """
    スコア関数の期待値がゼロであることを数値的に確認
    （正則条件を満たす分布族の重要な性質）
    """
    # サンプリング
    np.random.seed(42)
    x_samples = np.random.normal(mu, sigma, n_samples)
    
    # スコア関数の計算
    score_mu, score_sigma = score_function_gaussian(x_samples, mu, sigma)
    
    # 期待値の推定
    E_score_mu = np.mean(score_mu)
    E_score_sigma = np.mean(score_sigma)
    
    return E_score_mu, E_score_sigma

# 確認
mu, sigma = 2, 1.5
E_mu, E_sigma = verify_score_expectation_zero(mu, sigma)

print(f"正規分布 N({mu}, {sigma}²) でのスコア関数の期待値:")
print(f"  E[∂log p/∂μ] ≈ {E_mu:.6f}  (理論値: 0)")
print(f"  E[∂log p/∂σ] ≈ {E_sigma:.6f}  (理論値: 0)")
print("\n→ スコア関数の期待値は0に近い（正則条件を満たす）")
````

### 例3：ベルヌーイ分布のスコア関数

````python
def score_function_bernoulli(x, p):
    """
    ベルヌーイ分布 Ber(p) のスコア関数
    P(x|p) = p^x (1-p)^{1-x}
    log P = x log p + (1-x) log(1-p)
    ∂log P/∂p = x/p - (1-x)/(1-p) = (x-p)/(p(1-p))
    """
    return (x - p) / (p * (1 - p))

# 可視化
p_range = np.linspace(0.01, 0.99, 100)

fig, ax = plt.subplots(figsize=(10, 5))

# x=0 と x=1 の場合のスコア関数
score_x0 = score_function_bernoulli(0, p_range)
score_x1 = score_function_bernoulli(1, p_range)

ax.plot(p_range, score_x0, 'b-', linewidth=2, label='x=0: score = -1/(1-p)')
ax.plot(p_range, score_x1, 'r-', linewidth=2, label='x=1: score = 1/p')
ax.axhline(y=0, color='k', linestyle='--', alpha=0.5)

ax.set_xlabel('p', fontsize=12)
ax.set_ylabel('Score function ∂log P/∂p', fontsize=12)
ax.set_title('Bernoulli distribution score function', fontsize=11)
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_ylim(-10, 10)

plt.tight_layout()
plt.show()

# Fisher情報
print("\nベルヌーイ分布のFisher情報:")
print("I(p) = E[(∂log P/∂p)²] = 1/(p(1-p))")
for p in [0.1, 0.3, 0.5, 0.7, 0.9]:
    I_p = 1 / (p * (1 - p))
    print(f"  p = {p}: I(p) = {I_p:.4f}")
````

## 6. 他の概念との関係

### 6.1 カルマンフィルタとの接点

カルマンフィルタの状態更新を**接ベクトルの視点**から見る：
- 予測分布 → 更新分布への移動は「多様体上の曲線」
- この移動の「瞬間的な方向」が接ベクトル
- 更新量 $(\Delta\mu, \Delta\sigma)$ は接ベクトル的

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# パラメータ
mu_pred, sigma_pred = 0, 2.0   # 事前分布
mu_obs = 1.5                    # 観測値
sigma_obs = 0.8                 # 観測ノイズ

# カルマンゲインと更新
K = sigma_pred**2 / (sigma_pred**2 + sigma_obs**2)
mu_post = mu_pred + K * (mu_obs - mu_pred)
sigma_post = np.sqrt((1 - K) * sigma_pred**2)

# 左図：パラメータ空間での更新
ax1 = axes[0]

ax1.plot(mu_pred, sigma_pred, 'bo', markersize=12, label='Prior')
ax1.plot(mu_post, sigma_post, 'go', markersize=12, label='Posterior')
ax1.plot(mu_obs, sigma_obs, 'r^', markersize=12, label=f'Observation')

# 更新の矢印（接ベクトル的）
ax1.annotate('', xy=(mu_post, sigma_post), xytext=(mu_pred, sigma_pred),
             arrowprops=dict(arrowstyle='->', color='purple', lw=2))
ax1.text((mu_pred + mu_post)/2 - 0.3, (sigma_pred + sigma_post)/2 + 0.1,
         'Update vector\n(tangent-like)', fontsize=10, color='purple')

ax1.set_xlabel('μ', fontsize=12)
ax1.set_ylabel('σ', fontsize=12)
ax1.set_title('Kalman filter state update\nMovement on manifold', fontsize=11)
ax1.legend()
ax1.set_xlim(-1, 3)
ax1.set_ylim(0, 3)
ax1.grid(True, alpha=0.3)

# 右図：対応する分布の変化
ax2 = axes[1]
x = np.linspace(-5, 5, 200)

y_pred = gaussian_pdf(x, mu_pred, sigma_pred)
y_post = gaussian_pdf(x, mu_post, sigma_post)
y_obs = gaussian_pdf(x, mu_obs, sigma_obs)

ax2.plot(x, y_pred, 'b-', linewidth=2, label=f'Prior N({mu_pred:.1f}, {sigma_pred:.1f}²)')
ax2.plot(x, y_post, 'g-', linewidth=2, label=f'Posterior N({mu_post:.2f}, {sigma_post:.2f}²)')
ax2.plot(x, y_obs, 'r--', linewidth=2, label=f'Likelihood N({mu_obs:.1f}, {sigma_obs:.1f}²)')
ax2.axvline(x=mu_obs, color='r', linestyle=':', alpha=0.5, label='Observation')

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('p(x)', fontsize=12)
ax2.set_title('Corresponding distribution changes', fontsize=11)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"""\n【Kalman filter update】
- Prior: (μ, σ) = ({mu_pred:.1f}, {sigma_pred:.1f})
- Posterior: (μ, σ) = ({mu_post:.2f}, {sigma_post:.2f})
- Kalman gain K = {K:.3f}
- Update vector: (Δμ, Δσ) = ({mu_post-mu_pred:.2f}, {sigma_post-sigma_pred:.2f})
""")
````

### 6.2 Fisher情報行列との関係

**スコア関数の共分散がFisher情報行列**：
$$I(\theta) = E\left[\frac{\partial \log p}{\partial \theta} \cdot \frac{\partial \log p}{\partial \theta}^\top\right]$$

- スコア関数は接ベクトルの「関数表現」
- Fisher情報行列は接空間上の**内積（計量）**を定義
- 次節「リーマン計量」で詳しく扱う

### 6.3 後続の概念への橋渡し

| 概念 | 接ベクトルとの関係 |
|------|-------------------|
| リーマン計量（§1.5） | 接ベクトルの「長さ」を測る |
| Fisher情報行列 | スコア関数（接ベクトル）の共分散 |
| 自然勾配法 | 接空間上の勾配を計量で補正 |

## 7. 演習問題

### Q1. 接ベクトルの成分

点 $p = (\mu, \sigma) = (1, 2)$ での接ベクトル $v = 3(\partial/\partial\mu) - (\partial/\partial\sigma)$ は、どの方向を指すか？成分表示 $(a, b)$ で答えよ。

<details>
<summary><b>解答を見る</b></summary>

接ベクトル $v = 3(\partial/\partial\mu) - 1(\partial/\partial\sigma)$ の成分表示は：

$$(a, b) = (3, -1)$$

これは μ方向に+3、σ方向に-1 の移動を表す。つまり、μを増やしながらσを減らす方向（右下方向）を指す。

</details>

````python
# Q1の可視化
fig, ax = plt.subplots(figsize=(8, 6))

mu_p, sigma_p = 1, 2
ax.plot(mu_p, sigma_p, 'ko', markersize=12, label='p = (1, 2)')

# v = 3(∂/∂μ) - (∂/∂σ) → 成分は (3, -1)
v_mu, v_sigma = 3, -1
scale = 0.3  # 可視化用のスケール
ax.annotate('', xy=(mu_p + v_mu*scale, sigma_p + v_sigma*scale), xytext=(mu_p, sigma_p),
             arrowprops=dict(arrowstyle='->', color='red', lw=2))
ax.text(mu_p + v_mu*scale + 0.1, sigma_p + v_sigma*scale, 'v = (3, -1)', fontsize=11, color='red')

ax.set_xlabel('μ')
ax.set_ylabel('σ')
ax.set_title('Q1: Tangent vector v = 3(∂/∂μ) - (∂/∂σ)')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_xlim(-0.5, 3)
ax.set_ylim(0, 3)
plt.show()
````

### Q2. スコア関数の計算

ポアソン分布 $\text{Poisson}(\lambda)$ のスコア関数 $\partial \log P(x|\lambda)/\partial \lambda$ を計算せよ。

（ヒント: $P(x|\lambda) = \lambda^x e^{-\lambda} / x!$）

<details>
<summary><b>解答を見る</b></summary>

ポアソン分布: $P(x|\lambda) = \frac{\lambda^x e^{-\lambda}}{x!}$

対数尤度:
$$\log P(x|\lambda) = x \log \lambda - \lambda - \log(x!)$$

スコア関数:
$$\frac{\partial \log P}{\partial \lambda} = \frac{x}{\lambda} - 1$$

確認:
- $E[\partial \log P/\partial \lambda] = E[x/\lambda - 1] = \lambda/\lambda - 1 = 0$ ✓（スコア関数の期待値は0）

</details>

### Q3. カルマンフィルタとの対応

カルマンゲインが大きいとき、パラメータ空間での移動ベクトル（接ベクトル的）はどうなるか説明せよ。

<details>
<summary><b>解答を見る</b></summary>

**カルマンゲイン K が大きいとき:**

1. **K が大きい条件**:
   - $K = \sigma_{\text{pred}}^2 / (\sigma_{\text{pred}}^2 + \sigma_{\text{obs}}^2)$
   - $\sigma_{\text{obs}}$ が小さい（観測精度が高い）とき $K \approx 1$

2. **更新量（移動ベクトル）**:
   - $\Delta\mu = K(\mu_{\text{obs}} - \mu_{\text{pred}})$ → K大 ⇒ 観測値に大きく引っ張られる
   - $\sigma_{\text{post}} = \sqrt{(1-K)\sigma_{\text{pred}}^2}$ → K大 ⇒ σが大きく減少

3. **パラメータ空間での移動**:
   - μ方向: 観測値に向かって大きく移動
   - σ方向: 大きく減少（不確実性が減る）
   - 結果: 観測点に近づく方向への大きな移動

4. **情報幾何的解釈**:
   - 高精度観測 = 尤度関数が鋭い = Fisher的に「情報量が多い」
   - 多様体上で「遠く」まで移動できる

</details>
