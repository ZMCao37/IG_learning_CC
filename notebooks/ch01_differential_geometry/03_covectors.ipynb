# §1.3 余接ベクトルと1-形式 - 双対の世界

## 1. 概要

- **この節で学ぶこと**: 余接ベクトル（1-形式）の定義、接ベクトルとの双対関係、微分形式の基礎
- **前提知識**: 接ベクトル、線形汎関数、双対空間
- **情報幾何との関連**: **勾配と自然勾配の関係、e-接続とm-接続の双対性**

## 2. 直感的理解

### 接ベクトルと余接ベクトルの違い

- **接ベクトル**: 「どの方向に動くか」（速度ベクトル）
- **余接ベクトル**: 「各方向の動きをどう測るか」（測定器）

### 日常での例え

- 接ベクトル = 「東に3km、北に4km進む」（移動の仕方）
- 余接ベクトル = 「高度計」（どの移動に対しても高度変化を返す）

### 関数の微分 $df$

関数 $f$ の微分 $df$ は余接ベクトル（1-形式）：
$$df(v) = v(f) = \text{「$v$方向への$f$の変化率」}$$

### 情報幾何での重要性

- **通常の勾配** $\nabla f$: 接ベクトル空間の元
- **勾配の「原型」** $df$: 余接ベクトル空間の元
- 計量を使って $df \leftrightarrow \nabla f$ を対応させる
- **自然勾配** = Fisher計量を使った対応

## 3. 数学的定義

### 3.1 余接空間の定義

点 $p \in M$ における**余接空間** $T_p^* M$ は、接空間 $T_p M$ の双対空間：
$$T_p^* M = \{ \omega : T_p M \to \mathbb{R} \mid \omega \text{ は線形} \}$$

### 3.2 基底と双対基底

座標 $(x^1, \ldots, x^n)$ に対して：
- 接空間の基底: $\left\{ \frac{\partial}{\partial x^1}, \ldots, \frac{\partial}{\partial x^n} \right\}$
- 余接空間の基底: $\{ dx^1, \ldots, dx^n \}$

双対関係：
$$dx^i \left( \frac{\partial}{\partial x^j} \right) = \delta^i_j$$

### 3.3 1-形式の成分表示

余接ベクトル $\omega \in T_p^* M$ は：
$$\omega = \omega_i \, dx^i$$

接ベクトル $v = v^j \frac{\partial}{\partial x^j}$ との作用：
$$\omega(v) = \omega_i v^i$$

### 3.4 関数の微分

関数 $f: M \to \mathbb{R}$ の微分 $df$ は1-形式：
$$df = \frac{\partial f}{\partial x^i} dx^i$$

接ベクトルへの作用：
$$df(v) = v^i \frac{\partial f}{\partial x^i} = v(f)$$

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams['figure.figsize'] = (10, 6)

class Covector:
    """余接ベクトル（1-形式）のクラス"""
    def __init__(self, components):
        self.components = np.array(components)  # 下付き添字 ω_i
    
    def __call__(self, tangent_vector):
        """接ベクトルへの作用: ω(v) = ω_i v^i"""
        return np.dot(self.components, tangent_vector)
    
    def __repr__(self):
        terms = [f"{c}dx{i+1}" for i, c in enumerate(self.components) if c != 0]
        return " + ".join(terms) if terms else "0"

class TangentVector:
    """接ベクトルのクラス"""
    def __init__(self, components):
        self.components = np.array(components)  # 上付き添字 v^i
    
    def __repr__(self):
        terms = [f"{c}∂/∂x{i+1}" for i, c in enumerate(self.components) if c != 0]
        return " + ".join(terms) if terms else "0"
````

## 4. 可視化

### 4.1 接ベクトルと余接ベクトルの双対性

````python
def visualize_duality():
    """接ベクトルと余接ベクトルの双対関係を可視化"""
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # 左図：接ベクトルとしての解釈
    ax1 = axes[0]
    
    # 原点
    ax1.plot(0, 0, 'ko', markersize=10)
    
    # 基底ベクトル
    ax1.arrow(0, 0, 1, 0, head_width=0.1, head_length=0.05, fc='blue', ec='blue')
    ax1.arrow(0, 0, 0, 1, head_width=0.1, head_length=0.05, fc='red', ec='red')
    ax1.text(1.1, 0, r'$\partial/\partial x^1$', fontsize=14, color='blue')
    ax1.text(0.1, 1.1, r'$\partial/\partial x^2$', fontsize=14, color='red')
    
    # 接ベクトル v = 2∂/∂x¹ + 1∂/∂x²
    v = np.array([2, 1])
    ax1.arrow(0, 0, v[0], v[1], head_width=0.1, head_length=0.05, fc='green', ec='green', linewidth=2)
    ax1.text(v[0]+0.1, v[1]+0.1, r'$v = 2\partial_1 + \partial_2$', fontsize=12, color='green')
    
    ax1.set_xlim(-0.5, 3)
    ax1.set_ylim(-0.5, 2)
    ax1.set_aspect('equal')
    ax1.set_xlabel(r'$x^1$')
    ax1.set_ylabel(r'$x^2$')
    ax1.set_title('接ベクトル空間 $T_pM$\n（「動く方向」を表す）')
    ax1.grid(True, alpha=0.3)
    
    # 右図：余接ベクトル（等高線として）
    ax2 = axes[1]
    
    # 余接ベクトル ω = 1·dx¹ + 2·dx² の等高線
    omega = np.array([1, 2])
    x = np.linspace(-0.5, 3, 100)
    y = np.linspace(-0.5, 2, 100)
    X, Y = np.meshgrid(x, y)
    Z = omega[0] * X + omega[1] * Y
    
    # 等高線（余接ベクトルの「測定値」が一定の線）
    levels = np.arange(-1, 8, 0.5)
    cs = ax2.contour(X, Y, Z, levels=levels, colors='purple', alpha=0.6)
    ax2.clabel(cs, inline=True, fontsize=8)
    
    # ω の「向き」を示す矢印（勾配方向）
    ax2.arrow(1, 0.5, omega[0]*0.3, omega[1]*0.3, head_width=0.1, head_length=0.05, 
              fc='purple', ec='purple', linewidth=2)
    ax2.text(1.5, 1.2, r'$\omega = dx^1 + 2dx^2$', fontsize=12, color='purple')
    
    # 接ベクトル v を重ねて描画
    ax2.arrow(0, 0, v[0]*0.9, v[1]*0.9, head_width=0.1, head_length=0.05, 
              fc='green', ec='green', linewidth=2, alpha=0.7)
    
    # 作用の値
    omega_v = np.dot(omega, v)
    ax2.text(0.5, -0.3, f'$\\omega(v) = 1 \\cdot 2 + 2 \\cdot 1 = {omega_v}$', fontsize=12)
    
    ax2.set_xlim(-0.5, 3)
    ax2.set_ylim(-0.5, 2)
    ax2.set_aspect('equal')
    ax2.set_xlabel(r'$x^1$')
    ax2.set_ylabel(r'$x^2$')
    ax2.set_title('余接ベクトル $T_p^*M$\n（「測定器」として働く）')
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

visualize_duality()
````

### 4.2 関数の微分 $df$ としての1-形式

````python
def visualize_differential():
    """関数の微分 df を可視化"""
    fig = plt.figure(figsize=(14, 5))
    
    # 関数 f(x,y) = x² + y²
    def f(x, y):
        return x**2 + y**2
    
    def df(x, y):
        """df = 2x dx + 2y dy"""
        return np.array([2*x, 2*y])
    
    x = np.linspace(-2, 2, 50)
    y = np.linspace(-2, 2, 50)
    X, Y = np.meshgrid(x, y)
    Z = f(X, Y)
    
    # 左図：3D曲面
    ax1 = fig.add_subplot(121, projection='3d')
    ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.7)
    
    # 点 p = (1, 1) での接平面
    p = np.array([1, 1])
    f_p = f(p[0], p[1])
    df_p = df(p[0], p[1])
    
    # 接平面: z = f(p) + df_p · (x - p)
    x_plane = np.linspace(0, 2, 20)
    y_plane = np.linspace(0, 2, 20)
    X_plane, Y_plane = np.meshgrid(x_plane, y_plane)
    Z_plane = f_p + df_p[0]*(X_plane - p[0]) + df_p[1]*(Y_plane - p[1])
    ax1.plot_surface(X_plane, Y_plane, Z_plane, alpha=0.3, color='red')
    
    ax1.scatter([p[0]], [p[1]], [f_p], color='red', s=100)
    ax1.set_xlabel('x')
    ax1.set_ylabel('y')
    ax1.set_zlabel('f(x,y)')
    ax1.set_title(r'$f(x,y) = x^2 + y^2$' + '\n点(1,1)での接平面')
    
    # 右図：勾配ベクトル場
    ax2 = fig.add_subplot(122)
    
    # 等高線
    levels = [0.5, 1, 2, 3, 4]
    cs = ax2.contour(X, Y, Z, levels=levels, colors='blue', alpha=0.5)
    ax2.clabel(cs, inline=True, fontsize=10)
    
    # 勾配ベクトル場（df の「向き」）
    skip = 5
    ax2.quiver(X[::skip, ::skip], Y[::skip, ::skip], 
               2*X[::skip, ::skip], 2*Y[::skip, ::skip],
               color='red', alpha=0.6, scale=30)
    
    # 点 p での df
    ax2.plot(p[0], p[1], 'ko', markersize=10)
    ax2.arrow(p[0], p[1], df_p[0]*0.3, df_p[1]*0.3, head_width=0.1, 
              head_length=0.05, fc='green', ec='green', linewidth=2)
    ax2.text(p[0]+0.5, p[1]+0.5, f'$df|_p = {df_p[0]}dx + {df_p[1]}dy$', fontsize=11)
    
    ax2.set_xlabel('x')
    ax2.set_ylabel('y')
    ax2.set_title('df は各点で「最急上昇方向」を指す\n（等高線に直交）')
    ax2.set_aspect('equal')
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"点 p = (1, 1) での:")
    print(f"  f(p) = {f_p}")
    print(f"  df|_p = {df_p[0]}dx + {df_p[1]}dy")
    print(f"")
    print(f"接ベクトル v = ∂/∂x + ∂/∂y への作用:")
    v = np.array([1, 1])
    print(f"  df(v) = {np.dot(df_p, v)} （f の v 方向微分）")

visualize_differential()
````

### 4.3 計量による接空間と余接空間の同一視

````python
def visualize_metric_isomorphism():
    """計量による T_pM と T_p*M の同一視"""
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # 3つの計量を考える
    metrics = [
        (np.eye(2), 'ユークリッド計量\n$g_{ij} = \\delta_{ij}$'),
        (np.array([[4, 0], [0, 1]]), '異方性計量\n$g = \\text{diag}(4, 1)$'),
        (np.array([[1, 0], [0, 4]]), 'Fisher計量（例）\n$g = \\text{diag}(1, 4)$')
    ]
    
    # 余接ベクトル（同じ）
    omega = np.array([1, 1])  # df の成分
    
    for ax, (g, title) in zip(axes, metrics):
        # 計量の逆行列
        g_inv = np.linalg.inv(g)
        
        # 勾配ベクトル = g^{ij} ω_j
        grad = g_inv @ omega
        
        # 余接ベクトルの等高線を描画
        x = np.linspace(-2, 2, 100)
        y = np.linspace(-2, 2, 100)
        X, Y = np.meshgrid(x, y)
        Z = omega[0] * X + omega[1] * Y
        
        cs = ax.contour(X, Y, Z, levels=np.arange(-3, 4, 0.5), colors='purple', alpha=0.4)
        
        # 余接ベクトルを表す矢印（測定の方向）
        ax.arrow(0, 0, omega[0]*0.8, omega[1]*0.8, head_width=0.1, 
                 head_length=0.05, fc='purple', ec='purple', linewidth=2, alpha=0.5)
        ax.text(omega[0]*0.9, omega[1]*0.9 + 0.2, r'$\omega$', fontsize=12, color='purple')
        
        # 勾配ベクトル（計量で決まる）
        ax.arrow(0, 0, grad[0]*0.8, grad[1]*0.8, head_width=0.1, 
                 head_length=0.05, fc='green', ec='green', linewidth=2)
        ax.text(grad[0]*0.9 + 0.1, grad[1]*0.9, r'$\nabla f$', fontsize=12, color='green')
        
        # 計量楕円
        theta = np.linspace(0, 2*np.pi, 100)
        # g^{ij} v_i v_j = 1 の楕円
        eigenvalues, eigenvectors = np.linalg.eigh(g_inv)
        ellipse = eigenvectors @ np.diag(np.sqrt(eigenvalues)) @ np.array([np.cos(theta), np.sin(theta)])
        ax.plot(ellipse[0], ellipse[1], 'b--', linewidth=1.5, alpha=0.5, label='単位球（$g$計量）')
        
        ax.set_xlim(-2.5, 2.5)
        ax.set_ylim(-2.5, 2.5)
        ax.set_aspect('equal')
        ax.set_xlabel(r'$x^1$')
        ax.set_ylabel(r'$x^2$')
        ax.set_title(title + f'\n$\\nabla f = ({grad[0]:.1f}, {grad[1]:.1f})$')
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='gray', linewidth=0.5)
        ax.axvline(0, color='gray', linewidth=0.5)
    
    plt.tight_layout()
    plt.show()
    
    print("【ポイント】")
    print("- 同じ余接ベクトル ω = dx¹ + dx² でも")
    print("- 計量によって対応する接ベクトル（勾配）が異なる")
    print("- 自然勾配 = Fisher計量を使った対応")

visualize_metric_isomorphism()
````

### 4.4 情報幾何での応用：勾配と自然勾配

````python
def visualize_natural_gradient():
    """勾配と自然勾配の違いを可視化"""
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # 正規分布の場合: パラメータ (μ, σ)
    # Fisher計量 g = diag(1/σ², 2/σ²)
    
    def fisher_metric(sigma):
        return np.array([[1/sigma**2, 0], [0, 2/sigma**2]])
    
    # 損失関数の勾配（例）
    def loss_gradient(mu, sigma):
        # 例: KLダイバージェンスの勾配
        return np.array([mu - 1, sigma - 1])
    
    # パラメータ空間でのベクトル場
    mu_range = np.linspace(-1, 3, 10)
    sigma_range = np.linspace(0.5, 2.5, 10)
    MU, SIGMA = np.meshgrid(mu_range, sigma_range)
    
    # 左図：通常の勾配
    ax1 = axes[0]
    
    U = MU - 1  # dL/dμ
    V = SIGMA - 1  # dL/dσ
    
    ax1.quiver(MU, SIGMA, -U, -V, color='blue', alpha=0.7, scale=15)
    ax1.plot(1, 1, 'r*', markersize=20, label='最適点')
    
    ax1.set_xlabel('μ')
    ax1.set_ylabel('σ')
    ax1.set_title('通常の勾配降下\n$-\\nabla L = -(\\partial L/\\partial\\theta^i)$')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 右図：自然勾配
    ax2 = axes[1]
    
    U_nat = np.zeros_like(MU)
    V_nat = np.zeros_like(SIGMA)
    
    for i in range(len(mu_range)):
        for j in range(len(sigma_range)):
            sigma = SIGMA[j, i]
            grad = loss_gradient(MU[j, i], sigma)
            g = fisher_metric(sigma)
            g_inv = np.linalg.inv(g)
            nat_grad = g_inv @ grad
            U_nat[j, i] = nat_grad[0]
            V_nat[j, i] = nat_grad[1]
    
    ax2.quiver(MU, SIGMA, -U_nat, -V_nat, color='green', alpha=0.7, scale=15)
    ax2.plot(1, 1, 'r*', markersize=20, label='最適点')
    
    ax2.set_xlabel('μ')
    ax2.set_ylabel('σ')
    ax2.set_title('自然勾配降下\n$-I(\\theta)^{-1}\\nabla L$')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print("【観察】")
    print("- 通常勾配: μとσ方向で同じスケール")
    print("- 自然勾配: σが小さい領域でσ方向の移動が大きくなる")
    print("  → Fisher計量による『パラメータ空間の歪み』を補正")

visualize_natural_gradient()
````

## 5. 具体例

### 例1：双対基底の計算

````python
def demonstrate_dual_basis():
    """双対基底の関係を確認"""
    print("【双対基底の関係】")
    print()
    print("座標 (x¹, x²) に対して:")
    print("  接空間の基底: {∂/∂x¹, ∂/∂x²}")
    print("  余接空間の基底: {dx¹, dx²}")
    print()
    print("双対関係: dx^i(∂/∂x^j) = δ^i_j")
    print()
    
    # 具体的に確認
    basis_vectors = [np.array([1, 0]), np.array([0, 1])]  # ∂/∂x¹, ∂/∂x²
    basis_covectors = [Covector([1, 0]), Covector([0, 1])]  # dx¹, dx²
    
    print("確認:")
    for i, omega in enumerate(basis_covectors):
        for j, e in enumerate(basis_vectors):
            result = omega(e)
            print(f"  dx^{i+1}(∂/∂x^{j+1}) = {result}")
    
    print()
    print("【一般のベクトルへの作用】")
    v = TangentVector([3, 2])
    omega = Covector([1, 4])
    print(f"接ベクトル v = {v}")
    print(f"余接ベクトル ω = {omega}")
    print(f"作用 ω(v) = 1·3 + 4·2 = {omega(v.components)}")

demonstrate_dual_basis()
````

### 例2：座標変換での変換則

````python
def demonstrate_transformation():
    """座標変換での接ベクトル・余接ベクトルの変換則"""
    print("【座標変換での変換則】")
    print()
    print("座標変換 x → y = Ax の場合:")
    print()
    
    # 変換行列
    A = np.array([[2, 1], [0, 1]])  # y = Ax
    A_inv = np.linalg.inv(A)
    
    print(f"変換行列 A = ")
    print(A)
    print()
    
    # 接ベクトル（反変ベクトル）
    v_x = np.array([1, 1])  # x座標系での成分
    v_y = A @ v_x  # y座標系での成分: v^i' = (∂y^i'/∂x^j) v^j
    
    print(f"接ベクトル v:")
    print(f"  x座標系: v = {v_x}")
    print(f"  y座標系: v' = A·v = {v_y}")
    print(f"  （反変変換: v^i' = (∂y^i'/∂x^j) v^j）")
    print()
    
    # 余接ベクトル（共変ベクトル）
    omega_x = np.array([2, 3])  # x座標系での成分
    omega_y = A_inv.T @ omega_x  # y座標系での成分: ω_i' = (∂x^j/∂y^i') ω_j
    
    print(f"余接ベクトル ω:")
    print(f"  x座標系: ω = {omega_x}")
    print(f"  y座標系: ω' = (A⁻¹)ᵀ·ω = {omega_y}")
    print(f"  （共変変換: ω_i' = (∂x^j/∂y^i') ω_j）")
    print()
    
    # 作用は不変
    pairing_x = np.dot(omega_x, v_x)
    pairing_y = np.dot(omega_y, v_y)
    print(f"【作用は座標によらない】")
    print(f"  x座標系: ω(v) = {pairing_x}")
    print(f"  y座標系: ω'(v') = {pairing_y}")

demonstrate_transformation()
````

### 例3：勾配と余勾配（情報幾何への準備）

````python
def demonstrate_gradient_cogradient():
    """勾配と余勾配の関係"""
    print("【勾配と余勾配】")
    print()
    print("関数 f の微分 df は余接ベクトル（1-形式）")
    print("勾配 ∇f は接ベクトル")
    print()
    print("両者の関係（計量 g を使う）:")
    print("  df(v) = g(∇f, v)  for all v")
    print("  成分で: ∂f/∂x^i = g_{ij} (∇f)^j")
    print("  よって: (∇f)^i = g^{ij} ∂f/∂x^j")
    print()
    
    # 具体例: 正規分布のパラメータ空間
    sigma = 1.0
    g = np.array([[1/sigma**2, 0], [0, 2/sigma**2]])  # Fisher計量
    g_inv = np.linalg.inv(g)
    
    # 損失関数の微分（例）
    df = np.array([1, 1])  # ∂L/∂μ = 1, ∂L/∂σ = 1
    
    # ユークリッド勾配
    grad_euclidean = df  # δ^{ij} ∂f/∂x^j
    
    # 自然勾配（Fisher計量）
    grad_natural = g_inv @ df  # g^{ij} ∂f/∂x^j
    
    print(f"σ = {sigma} での計算:")
    print(f"  df = ({df[0]}, {df[1]})  [余接ベクトル/共変成分]")
    print(f"  ユークリッド勾配: {grad_euclidean}  [反変成分, δ計量]")
    print(f"  自然勾配: {grad_natural}  [反変成分, Fisher計量]")
    print()
    print("【情報幾何の視点】")
    print("- df は『本質的な』勾配情報（座標系によらない）")
    print("- ∇f は計量の選び方に依存")
    print("- 自然勾配 = Fisher計量で df を接ベクトルに変換したもの")

demonstrate_gradient_cogradient()
````

## 6. 他の概念との関係

### 前の節との繋がり
- **線形代数 (§0.1)**: 双対空間、線形汎関数の概念
- **接ベクトル (§1.2)**: 余接空間は接空間の双対

### 次の節への接続
- **ベクトル場 (§1.4)**: 1-形式場（微分形式）
- **リーマン計量 (§1.5)**: 計量による $TM \leftrightarrow T^*M$ の同一視

### 情報幾何での重要性

| 概念 | 接空間 $T_pM$ | 余接空間 $T_p^*M$ |
|:---|:---|:---|
| 基底 | $\partial/\partial\theta^i$ | $d\theta^i$ |
| 成分 | 上付き $v^i$（反変） | 下付き $\omega_i$（共変） |
| 例 | 勾配ベクトル $\nabla f$ | 微分 $df$ |
| 変換 | $(\partial y/\partial x)$ | $(\partial x/\partial y)^\top$ |

### 双対座標との関係（情報幾何）

- **自然パラメータ** $\theta$ と **期待値パラメータ** $\eta$ は双対座標
- $\theta$-座標での勾配 $\leftrightarrow$ $\eta$-座標での勾配
- これが e-接続と m-接続の双対性につながる

### Q2. 関数の微分

関数 $f(x, y) = xy^2$ の点 $(2, 3)$ における微分 $df$ を求めよ。

<details>
<summary>解答を見る</summary>

$$df = \frac{\partial f}{\partial x}dx + \frac{\partial f}{\partial y}dy = y^2 dx + 2xy\, dy$$

点 $(2, 3)$ で: $df = 9\,dx + 12\,dy$

</details>

````python
# Q2検証
x, y = 2, 3
df_x = y**2  # ∂f/∂x
df_y = 2*x*y  # ∂f/∂y
print(f"df|_(2,3) = {df_x}dx + {df_y}dy")
````

### Q3. 自然勾配の計算

Fisher計量 $g = \text{diag}(1, 4)$ で、$df = 2dx + dy$ に対応する勾配ベクトル $\nabla f$ を求めよ。

<details>
<summary>解答を見る</summary>

$$g^{-1} = \text{diag}(1, 1/4)$$

$$(\nabla f)^i = g^{ij} (df)_j$$

$$\nabla f = \begin{pmatrix} 1 & 0 \\ 0 & 1/4 \end{pmatrix} \begin{pmatrix} 2 \\ 1 \end{pmatrix} = \begin{pmatrix} 2 \\ 1/4 \end{pmatrix}$$

</details>

````python
# Q3検証
g = np.diag([1, 4])
g_inv = np.linalg.inv(g)
df = np.array([2, 1])
grad_f = g_inv @ df
print(f"∇f = {grad_f}")
````
